```python
import spacy
import spacy_experimental
```

# First Example without Heads

#### Vortrainiertes englisches Coreference Modell wird der Variable nlp zugewiesen

```python
from spacy_experimental.coref.coref_component import DEFAULT_COREF_MODEL
from spacy_experimental.coref.coref_util import DEFAULT_CLUSTER_PREFIX
nlp = spacy.load("en_core_web_trf")
config={
    "model": DEFAULT_COREF_MODEL,
    "span_cluster_prefix": DEFAULT_CLUSTER_PREFIX,
}
nlp.add_pipe("experimental_coref", config=config)

```

#### Beispieltext wird durch das geladene Modell verarbeitet

```python
doc = nlp("The cats were startled by the dog as it growled at them.")
```

#### doc.spans ist ein Attribut des Doc-Objekts, das es ermöglicht, zusammenhängenden Wörtern oder Tokens im Text zu speichern.
#### In der Grundfunktion werden ganze Koreferenzcluster umfasst. Durch die Berücksichtigung gesamter Cluster wird der Kontext besser erfasst.Ein Head kann aus mehreren Teilen bestehen (Bsp.: Mr. George Smith, Head kann George oder Smith sein)

```python
doc.spans
```

# Second Example with Heads

#### Dieser Code schafft es sowohl einzelne Heads als auch Cluster zu filtern.

```python
nlp = spacy.load("en_core_web_sm")
nlp_coref = spacy.load("en_coreference_web_trf")

# use replace_listeners for the coref components
nlp_coref.replace_listeners("transformer", "coref", ["model.tok2vec"])
nlp_coref.replace_listeners("transformer", "span_resolver", ["model.tok2vec"])

# we won't copy over the span cleaner
nlp.add_pipe("coref", source=nlp_coref)
nlp.add_pipe("span_resolver", source=nlp_coref)

doc = nlp("The cats were startled by the dog as it growled at them.") 

# print(doc.spans)
```
